# ThingSpeak Moving Average

https://thingspeak.mathworks.com/channels/1470727

```matlab
data = thingSpeakRead(1470727,Fields=1,NumPoints=100,OutputFormat='TimeTable');
head(data)
```

```matlab
data(1:10,:)
```

```matlab
data.MovingAverage = movmean(data.FilteredResults, 3);
```

```matlab
t = data.Timestamps;
x = data.FilteredResults;
y = data.MovingAverage;
plot(t,x,t,y)
```
